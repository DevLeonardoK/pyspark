## Exercício 5 — Substituir uma View
Use `CREATE OR REPLACE VIEW` para atualizar `vw_produtos_caros`  
para também incluir a coluna `estoque` no resultado.  
Consulte a view atualizada.


```python
from pyspark.sql.functions import col
```

```python

%sql
-- Exercício 1 — use %sql no Databricks ou spark.sql() aqui

CREATE TABLE if not exists produtos (
    id INT,
    nome VARCHAR(255),
    categoria VARCHAR(100),
    preco DOUBLE,
    estoque INT
) USING DELTA;

INSERT INTO produtos VALUES
(1, 'Produto 1', 'Eletrônico', 100.50, 2),
(2, 'Produto 2', 'Eletrônico', 250.00, 5),
(3, 'Produto 3', 'Informática', 1500.99, 10),
(4, 'Produto 4', 'Móveis', 799.90, 3),
(5, 'Produto 5', 'Acessórios', 49.99, 20);
```

```python
# Exercício 2
df2 = spark.read.table("produtos")
df2 = df2.select("nome").filter(col("preco")>1000).createOrReplaceTempView("vw_produtos_caros")

spark.catalog.listTables()
spark.sql("""select * from vw_produtos_caros""").show()
```

```python
# Exercício 3
df3 = spark.read.table("produtos")
df3.filter(col("estoque")<5).createOrReplaceTempView("vw_estoque_baixo")
spark.sql("""select * from vw_estoque_baixo""").show()
```

```python
%sql
--Exercício 4

create table if not exists clientes_teste (
    id bigint,
    nome varchar(255),
    saldo decimal(10,2),
    ativo boolean,
    criado_em timestamp
);

INSERT INTO clientes_teste VALUES
(1, 'João Silva', 1500.75, true,  TIMESTAMP('2026-01-10 08:30:00')),
(2, 'Maria Souza', 3200.50, false, TIMESTAMP('2026-02-15 14:45:00')),
(3, 'Carlos Lima', 850.00, true,   TIMESTAMP('2026-03-20 09:15:00'));

```

```python
# Exercício 5
df2 = spark.read.table("produtos")
df2 = df2.select("nome", "estoque").filter(col("preco")>1000).createOrReplaceTempView("vw_produtos_caros")

spark.sql("""select * from vw_produtos_caros""").show()
```
