```python
# Exercício 1
path_vendas = "/Workspace/Users/leocodedev@outlook.com/pyspark/notebooks/exercise/data/vendas.csv"
ex1_vendas_df = spark.read.csv(path_vendas, header=True, inferSchema=True)
ex1_vendas_df.show(5)
ex1_vendas_df.printSchema()
```

```python
# Exercício 2
from pyspark.sql.types import StructType, StringType, DateType, DoubleType, StructField, IntegerType
path_vendas = "/Workspace/Users/leocodedev@outlook.com/pyspark/notebooks/exercise/data/vendas.csv"
ex2_vendas_schema = StructType(
    [
    StructField("data", DateType(), nullable=True), 
     StructField("produto", StringType(), nullable=False),
     StructField("categoria", StringType(), nullable=True),
     StructField("valor", DoubleType(), nullable=False),
     StructField("quantidade", IntegerType(), nullable=False),
     StructField("regiao", StringType(), nullable=True),
     StructField("vendedor", StringType(), nullable=True)
    ]
    )
ex2_vendas_df = spark.read.csv(path_vendas, header=True, schema=ex2_vendas_schema)
ex2_vendas_df.printSchema()
ex2_vendas_df.show(5)
```

```python
# Exercício 3
ex3_vendas_df = ex2_vendas_df
rows = ex3_vendas_df.count()
ex3_vendas_df.describe("valor","quantidade")
print(rows)
```

```python
# Exercício 4
from pyspark.sql.functions import col, trim, lower,desc
ex4_vendas_df = ex3_vendas_df
ex4_vendas_df_regiao_sul = ex4_vendas_df.select("data","produto","vendedor").filter(lower(trim(col("regiao"))) =="sul").orderBy(col("valor").desc())
display(ex4_vendas_df_regiao_sul)
```

```python
# Exercício 5

ex5_vendas_df = ex4_vendas_df
ex5_vendas_df = ex5_vendas_df.withColumn("valor_total", col("valor") * col("quantidade")).select("produto","valor","quantidade","valor_total")
display(ex5_vendas_df)

```
