## Exercício 2 — COPY INTO
Use `COPY INTO` para ingerir `clientes.csv` em uma nova tabela Delta chamada `clientes_delta`:  
```sql
COPY INTO clientes_delta
FROM '/FileStore/tables/clientes.csv'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true', 'inferSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
```
Execute duas vezes e observe: `COPY INTO` é **idempotente** (não duplica dados).

## Exercício 4 — MERGE (CDC / Upsert)
Crie um DataFrame com 2 clientes existentes (com dados atualizados) e 1 cliente novo.  
Use `MERGE INTO` para fazer upsert na tabela `clientes_delta`:  
- Se o `id` já existir → atualize o `nome` e `email`  
- Se não existir → insira o novo registro

```sql
MERGE INTO clientes_delta AS alvo
USING novos_clientes AS fonte
ON alvo.id = fonte.id
WHEN MATCHED THEN UPDATE SET ...
WHEN NOT MATCHED THEN INSERT ...
```
