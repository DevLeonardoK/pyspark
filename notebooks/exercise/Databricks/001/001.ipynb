```python
import pyspark.sql.functions as f
```

```python
#Ex1
data = [(1,"Notebook",3500), (2,"Mouse", 150), (3, "Teclado", 250)]
columns = ["id", "produto", "preco"]

df_ex1 = spark.createDataFrame(data = data, schema = columns)

df_ex1 = df_ex1.withColumn("desconto", f.lit(None))

display(df_ex1)
```

```python
#Ex2
data = [(1, "Ana", 25),(2, None, 30),(3, "Carlos", None),(4, "João", 40)]
columns = ["id","nome","idade"]

df_ex2 = spark.createDataFrame(data = data, schema = columns)
display(df_ex2)
df_ex2 = df_ex2.dropna()
display(df_ex2)
```

```python
#Ex3
data = [(1, "Ana", "TI"),(2, "Bruno", None),(3, "Carlos", None),(4, "Diana", "RH")]
columns = ["id", "nome", "setor"]

df_ex3 = spark.createDataFrame(data = data, schema = columns)
display(df_ex3)
df_ex3 = df_ex3.fillna("Não informado", subset=["setor"])
display(df_ex3)
```

```python
#Ex4
data = [("Ana", "TI", 5000),("Bruno", "RH", 3000),("Carlos", "Financeiro", 2000),("Diana", "TI", 4500)]
columns = ["nome", "departamento", "salario"]

df_ex4 = spark.createDataFrame(data = data, schema = columns)
display(df_ex4)
df_ex4 = df_ex4.withColumn("faixa_salarial", f.when(f.col("salario") > 4000, "Alta").when(f.col("salario") >=2500, "Média").otherwise("Baixa"))
display(df_ex4)
```

```python
#Ex5
df_ex5 = df_ex4
df_ex5_media_salario_departamento = df_ex5.groupBy(f.col("departamento")).agg(f.avg(f.col("salario")).alias("media_salarial_departamento")).orderBy(f.col("media_salarial_departamento").asc())
display(df_ex5_media_salario_departamento)
```
