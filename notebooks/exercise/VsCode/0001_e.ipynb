  data = [("Leonardo", "Porto Alegre"), ("Julia", "São Paulo"), ("Elaine", "Curitiba")]
  columns = ["Nome", "Cidade"]

  Exercício 2 — filter com string
  Usando o mesmo DataFrame do exercício 1, filtre apenas as pessoas cuja cidade seja "São Paulo".

  Exercício 3 — filter + select combinados
  Crie um DataFrame com Nome, Idade e Salario:
  data = [("Ana", 25, 3000), ("Bruno", 17, 1500), ("Carlos", 32, 7000), ("Diana", 15, 800)]
  columns = ["Nome", "Idade", "Salario"]
  Filtre apenas os maiores de 18 anos e exiba somente Nome e Salario.

  ---
  Exercício 4 — orderBy decrescente
  Usando o DataFrame do exercício 3, ordene todos os registros por Salario em ordem decrescente e exiba o resultado completo.

  ---
  Exercício 5 — encadeamento completo
  Com o mesmo DataFrame do exercício 3, em uma única linha encadeada:
  - Filtre quem tem Salario > 2000
  - Ordene por Nome em ordem ascendente
  - Exiba apenas as colunas Nome e Idade

  ---

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,asc,desc
```

```python
spark = SparkSession.builder.appName("0001_e").master("local[*]").getOrCreate()
spark.active()
```

```python
#exericio 1
data_1 = [
    ("Leonardo","Porto Alegre"), 
    ("Julia","São Paulo"),
    ("Elaine", "Curitiba"),
]

columns_1 = ["Nome", "Cidade"]

df_1 = spark.createDataFrame(data=data_1, schema=columns_1)
exercicio_1 = df_1.select("Cidade")
exercicio_1.show()
```

```python
#exercicio2 2
exercicio_2 = df_1.filter(col("Cidade") == "São Paulo")
exercicio_2.show()
```

```python
#exercicio 3
data_3 = [("Ana",25,3000),("Bruno",17,1500),("Carlos",32, 7000), ("Diana",15,800)]
columns_3 = ["Nome","Idade","Salário"]

df_3 = spark.createDataFrame(data=data_3,schema=columns_3)
exercicio_3 = df_3.filter(col("Idade") > 18).select("Nome","Salário").show()
```

```python
#exercicio 4
df_4 = df_3
exercicio_4 = df_4.orderBy(desc(col("Salário"))).show()
```

```python
df_5 = df_3
exercicio_5 = df_5.select("Nome","Idade").filter(col("Salário") > 2000).orderBy(asc(col("Nome"))).show()
```
