Manipulação de dados com PySpark

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("0003_spark").master("local[*]").getOrCreate()
```

```python
#Leitura de CSV
df_csv = spark.read.csv("notebooks/data/airtravel.csv", header=True, inferSchema=True)
df_csv.show()
```

```python
#Leitura Parquet

df_parquet = spark.read.parquet("notebooks/data/mtcars.parquet")
df_parquet.show()
```

```python
#Escrever dados CSV

df_csv.write.mode("overwrite").csv("/home/jovyan/notebooks/data/airtravel_2.csv")
```

```python
#Escrever dados PARQUET

df_parquet.write.mode("overwrite").parquet("/home/jovyan/notebooks/data/mtcars_2.parquet")
```
