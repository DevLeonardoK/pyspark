Transformações X Ações Pyspark

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import asc, desc, col
```

```python
spark = SparkSession.builder.appName("spark_0002").master("local[*]").getOrCreate()
spark.active()
```

```python
data = [("John", 28), ("Anna",24), ("Mike", 35)]
columns = ["Nome", "Idade"]

df = spark.createDataFrame(data=data, schema=columns)
```

```python
df = df.withColumn("Idade - Mais 5 anos", col("Idade")+5)
df.show()
```

```python
#Transformação
transformed_df = df.select("Nome","Idade").filter(col("Idade") > 15)
```

```python
#Ação
transformed_df.show()
```
