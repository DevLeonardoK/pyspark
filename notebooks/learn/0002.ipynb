Transformações X Ações Pyspark

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import asc, desc, col, round
```

```python
spark = SparkSession.builder.appName("spark_0002").master("local[*]").getOrCreate()
spark.active()
```

```python
data = [("John", 28), ("Anna",24), ("Mike", 35)]
columns = ["Nome", "Idade"]

df = spark.createDataFrame(data=data, schema=columns)
```

```python
df = df.withColumn("Idade - Mais 5 anos", col("Idade")+5)
df.show()
```

```python
#Transformação
transformed_df = df.select("Nome","Idade").filter(col("Idade") > 15)
```

```python
#Ação
transformed_df.show()
```

```python
#Exemplo Prático - Montagem dados
data = [
    ("Johnshon","Vendas", 3000),
    ("Anna","Marketing", 4500),
    ("Mike", "Vendas", 3500),
    ("Sara", "Marketing", 4000),
    ("João", "Vendas", 3000)
]

columns = ["Nome", "Setor", "Salário"]

df = spark.createDataFrame(data=data, schema=columns)
df.show()
```

```python
#Exemplo Prático - Transformação & Ação

#Transformação
#Filtro - Funcionários (Vendas)
funcionarios_vendas = df.filter(col("Setor") == "Vendas")

#Ação
#show
funcionarios_vendas.show()
```

```python
#Exemplo Prático - Transformação & Ação

#Tranformação
#Média salarial
media_salarial = df.groupBy(col("Setor")).avg("Salário")

#Ação
#Show
media_salarial.select("Setor", round("avg(Salário)",2).alias("Salário")).orderBy(asc(col("Salário"))).show()
```
