```python
from pyspark.sql.functions import col, lit, avg, sum
```

```python
data = [
    (1, "Ana", 25),
    (2, "", 30),
    (3, "Carlos", None),
    (4, None, ""),
    (5, "João", 40)
]

schema = ["id", "nome", "idade"]

df_apoio = spark.createDataFrame(data, schema)
```

```python
#Usando lit - cria coluna literal

df_apoio = df_apoio.withColumn("nula", lit(None))
display(df_apoio)
```

```python
df_apoio = df_apoio.dropna()
#dropna vai dropar linhas com valores nulos
display(df_apoio)
```

```python
#Preenchendo valores ausentes

from pyspark.sql.functions import col, lit, avg, sum
data = [
    (1, "", 25),
    (2, "", 30),
    (3, "Carlos", None),
    (4, None, ""),
    (5, "João", 40)
]

schema = ["id", "nome", "idade"]

df_preencher = spark.createDataFrame(data, schema)

#Preencher dados nulos - fillna
df_preencher = df_preencher.fillna("Desconhecido", subset=["nome"])

display(df_preencher)
```

```python
#Parte prática
from pyspark.sql.functions import col, lit, avg, sum, when

data = [
    (1, "Ana", "TI", 3000),
    (2, None, "RH", 2500),
    (3, "Carlos", None, None),
    (4, None, "Financeiro", 5000),
    (5, "João", "TI", 4000)
]

schema = ["id", "nome", "departamento", "salario"]

df_pratica = spark.createDataFrame(data, schema)
```

```python
avg_salario = df_pratica.agg(avg(col("salario"))).collect()[0][0] #----.collect() - retorna todos os dados na memoria local, cuidado, muito pesado, melhor usar display ou show
display(avg_salario)
```

```python
df_pratica = df_pratica.fillna(avg_salario, subset=["salario"])
display(df_pratica)
```

```python
df_pratica = df_pratica.withColumn("nome", when(col("nome").isNull(), "Desconhecido").otherwise(col("nome"))) #withColumn deve ser string e não objeto column
display(df_pratica)
```

```python
df_pratica = df_pratica.dropna()
display(df_pratica)
```
