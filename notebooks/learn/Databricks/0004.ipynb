Funções no pyspark = 
col - Selecionar coluna
lit - Criar valor literal
when - Cria lógica condicional
groupBy - Agrupa os dados por

```python
import pyspark.sql.functions as F
```

```python
data = [("John","Vendas",3000), ("Anna", "Marketing",4500), ("Mike", "Vendas",3500),("Sara","Marketing",4000)]
columns = ["Nome", "Departamento", "Salário"]

df = spark.createDataFrame(data = data, schema = columns)
```

```python
display(df)
```

```python
df = df.withColumn("salario_categoria", F.when(F.col("Salário") > 3000, "Alto").otherwise("Baixo"))
display(df)
```

```python
groupby_avg = df.groupBy(F.col("Departamento")).agg(F.avg(F.col("Salário")))
display(groupby_avg)
```

```python
groupby_sum = df.groupBy(F.col("Departamento")).agg(F.sum(F.col("Salário")).alias("salario_soma")).orderBy(F.col("salario_soma"),ascending=False)
display(groupby_sum)
```

```python
data_null = [("Leonardo", "", 19), ("Elaine", "Cozinheira", ""), ("Rafaela", "Comerciante", None), (None, "Gerente", 35)]
columns = ["Nome", "Profissão", "Idade"]

df_null = spark.createDataFrame(data = data_null, schema = columns)
display(df_null)
```

```python
df_null.printSchema()

df_null = df_null.withColumn("Idade", F.when(F.col("Idade") == "", None).otherwise(F.col("Idade")))
df_null = df_null.withColumn("Profissão", F.when(F.col("Profissão") == "", None).otherwise(F.col("Profissão")))

df_null = df_null.withColumn("Idade", F.col("Idade").cast("double"))

media_idade = df_null.agg(F.avg(F.col("Idade"))).first()[0]

df_null = df_null.fillna(value = media_idade, subset = ["Idade"])

df_null = df_null.fillna(value = "Desconhecido", subset = ["Profissão", "Nome"])

display(df_null)

#dropna = Remove linha que tem colunas com valores null / dropna(how="all") remove somente se todas as colunas da linha estiverem null
#fillna = Faz replace dos valores nulos por selecionados, usar value =  e subset = para colunas
```
