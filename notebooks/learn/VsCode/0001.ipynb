Funções iniciais = select, filter, orderBy

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,asc,desc
```

```python
spark = SparkSession.builder.master("local[*]").appName("spark_0001").getOrCreate()
spark.active()
```

```python
data = [("Leonardo",20),("Julia",18),("Elaine",40)]
columns = ["Nome","Idade"]
```

```python
df = spark.createDataFrame(data=data, schema=columns)
```

```python
df.show()
```

```python
df.select("Nome").show()
```

```python
df.select("Nome","Idade").show()
```

```python
df_filter_idade = df.filter(df.Idade >= 18)
df_filter_idade.show()
```

```python
df.filter(col("idade") > 15).orderBy(asc(col("Nome"))).show()
```

```python
df_orderby_idade = df.orderBy(df.Idade.asc())
df_orderby_idade.show()
```
