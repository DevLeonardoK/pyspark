```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.master("local[*]").getOrCreate()
spark.active()
```

```python
data = [("Leonardo",19),("John",30),("Julia",18)]
columns = ["Nome","Idade"]
```

```python
df = spark.createDataFrame(data=data,schema=columns)
df.show()
```

```python
df.createOrReplaceTempView("tabela_usuarios")
```

```python
sql = spark.sql
```

```python
sql("SELECT * FROM TABELA_USUARIOS").show()
```
